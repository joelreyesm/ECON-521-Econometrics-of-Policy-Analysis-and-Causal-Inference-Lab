# Part C. LATE in a world where you know the truth (30 points)

In the Card data we never see potential outcomes, so we cannot check *whose* effect IV recovers. Here
we can. The DGP below is a stylized version of the scholarship story, built so that, as Card argues,
people who find college costly are also the people with the highest returns.

### The DGP

For $i = 1, \dots, N$ with $N = 200{,}000$:

$$
\begin{aligned}
V_i &\sim \text{Uniform}(0,1) && \text{resistance to college (cost, distance, credit constraints); unobserved}\\
\varepsilon_i &\sim N(0,\, 0.3^2)\\
Y_i(0) &= 6.0 - 0.4\,V_i + \varepsilon_i && \text{log wage without college}\\
\tau_i &= 0.10 + 0.30\,V_i && \text{individual return to college}\\
Y_i(1) &= Y_i(0) + \tau_i\\
Z_i &\in \{0, 1, 2\} \text{ with probabilities } (0.5,\ 0.3,\ 0.2) && \text{0 = no offer, 1 = partial scholarship, 2 = full scholarship}\\
D_i &= \mathbf{1}\{V_i < p(Z_i)\}, \qquad p(0) = 0.2,\ p(1) = 0.6,\ p(2) = 0.9 && \text{college attendance}\\
Y_i &= D_i\,Y_i(1) + (1 - D_i)\,Y_i(0)
\end{aligned}
$$

$Z$ is drawn independently of $(V_i, \varepsilon_i)$. Use seed **521**. Your numbers will differ
slightly from the answer key depending on the order of your draws; answers within ±0.02 get full
credit.

Since you simulate the potential outcomes, you can compute any **true** average effect directly from
$\tau_i$. Do so whenever a question asks for a "true" effect.
